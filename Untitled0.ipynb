```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf

data = yf.download(
    "SPY",
    start="2018-01-01",
    end="2026-01-01",
    auto_adjust=True,
    progress=False
)

close = data["Close"]

if isinstance(close, pd.DataFrame):
    close = close.iloc[:, 0]

close = close.dropna().rename("Close")

print(close.head())
print("Number of observations:", len(close))

close.plot(figsize=(12, 5), title="SPY Historical Price")
plt.ylabel("Price (USD)")
plt.show()
```

```python

# Calculate daily returns
returns = close.pct_change().dropna()

# Display the first few observations
print(returns.head())

# Summary statistics
print("\nSummary statistics:")
print(returns.describe())

# Plot daily returns
returns.plot(
    figsize=(12, 5),
    title="SPY Daily Returns"
)

plt.ylabel("Daily Return")
plt.axhline(0, linewidth=1)
plt.show()
```

```python

# Calculate moving averages
ma20 = close.rolling(window=20).mean()
ma50 = close.rolling(window=50).mean()

# Plot price and moving averages
plt.figure(figsize=(12, 6))
plt.plot(close.index, close, label="SPY Price", alpha=0.6)
plt.plot(ma20.index, ma20, label="20-day Moving Average")
plt.plot(ma50.index, ma50, label="50-day Moving Average")

plt.title("SPY Price with Moving Averages")
plt.xlabel("Date")
plt.ylabel("Adjusted Price (USD)")
plt.legend()
plt.show()
```

```python

# 1. Create the trading signal
signal = (ma20 > ma50).astype(int)

# 2. Use yesterday's signal for today's return
position = signal.shift(1).fillna(0)

# 3. Calculate strategy returns
strategy_returns = position * returns

# 4. Calculate buy-and-hold returns
buy_hold_returns = returns

# 5. Calculate cumulative growth of $1
strategy_growth = (1 + strategy_returns).cumprod()
buy_hold_growth = (1 + buy_hold_returns).cumprod()

# 6. Plot the comparison
plt.figure(figsize=(12, 6))
plt.plot(
    strategy_growth.index,
    strategy_growth,
    label="Moving Average Strategy"
)
plt.plot(
    buy_hold_growth.index,
    buy_hold_growth,
    label="Buy and Hold"
)

plt.title("Strategy vs Buy and Hold")
plt.xlabel("Date")
plt.ylabel("Growth of $1")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python

def evaluate_strategy(daily_returns, name):
    daily_returns = daily_returns.dropna()

    # Cumulative growth of $1
    growth = (1 + daily_returns).cumprod()

    # Total return
    total_return = growth.iloc[-1] - 1

    # Annualised volatility: approximately 252 trading days per year
    annual_volatility = daily_returns.std() * np.sqrt(252)

    # Maximum drawdown
    running_max = growth.cummax()
    drawdown = growth / running_max - 1
    max_drawdown = drawdown.min()

    return pd.Series({
        "Total Return": total_return,
        "Annualised Volatility": annual_volatility,
        "Maximum Drawdown": max_drawdown
    }, name=name)


results = pd.concat([
    evaluate_strategy(strategy_returns, "Moving Average"),
    evaluate_strategy(buy_hold_returns, "Buy and Hold")
], axis=1)

print((results * 100).round(2).astype(str) + "%")
```

```python

# Simplified annualised Sharpe ratio
# Assumes a risk-free rate of 0%

def sharpe_ratio(daily_returns):
    daily_returns = daily_returns.dropna()
    return (
        daily_returns.mean()
        / daily_returns.std()
        * np.sqrt(252)
    )

print(
    "Moving Average Sharpe Ratio:",
    round(sharpe_ratio(strategy_returns), 3)
)

print(
    "Buy and Hold Sharpe Ratio:",
    round(sharpe_ratio(buy_hold_returns), 3)
)
```

```python

# Illustrative transaction cost: 0.1% per position change
cost_rate = 0.001

# Position changes: 1 when the holding switches between cash and SPY
position_change = position.diff().abs()

# Include the initial entry cost if the first position is invested
if len(position_change) > 0:
    position_change.iloc[0] = abs(position.iloc[0])

# Net strategy returns after transaction costs
strategy_returns_net = (
    position * returns
    - position_change * cost_rate
)

print("Gross Sharpe Ratio:",
      round(sharpe_ratio(strategy_returns), 3))

print("Net Sharpe Ratio:",
      round(sharpe_ratio(strategy_returns_net), 3))

print("Gross Total Return:",
      round(((1 + strategy_returns).prod() - 1) * 100, 2), "%")

print("Net Total Return:",
      round(((1 + strategy_returns_net).prod() - 1) * 100, 2), "%")
```

```python

# 1. Split the data into training and test periods
split_date = "2023-01-01"

train_strategy = strategy_returns_net[
    strategy_returns_net.index < split_date
]
test_strategy = strategy_returns_net[
    strategy_returns_net.index >= split_date
]

train_buy_hold = buy_hold_returns[
    buy_hold_returns.index < split_date
]
test_buy_hold = buy_hold_returns[
    buy_hold_returns.index >= split_date
]

# 2. Evaluate performance
def evaluate_period(daily_returns):
    daily_returns = daily_returns.dropna()

    growth = (1 + daily_returns).cumprod()
    total_return = growth.iloc[-1] - 1
    annual_volatility = daily_returns.std() * np.sqrt(252)

    running_max = growth.cummax()
    drawdown = growth / running_max - 1
    max_drawdown = drawdown.min()

    sharpe = (
        daily_returns.mean()
        / daily_returns.std()
        * np.sqrt(252)
    )

    return pd.Series({
        "Total Return (%)": total_return * 100,
        "Annual Volatility (%)": annual_volatility * 100,
        "Maximum Drawdown (%)": max_drawdown * 100,
        "Sharpe Ratio": sharpe
    })

# 3. Compare both strategies in each period
results = pd.DataFrame({
    "Train: Moving Average": evaluate_period(train_strategy),
    "Train: Buy and Hold": evaluate_period(train_buy_hold),
    "Test: Moving Average": evaluate_period(test_strategy),
    "Test: Buy and Hold": evaluate_period(test_buy_hold)
})

print(results.round(2))
```
